# Showcase Notebook 02: HPC & Performance Benchmarking Story
This Jupyter notebook illustrates matrix operations and timing tracking for high-performance computing development stories in `provBookR`.

```r
# 1. Matrix Initialization and SVD Decompositions
matrix_size <- 400
n_iterations <- 5
logs <- vector('list', n_iterations)

for (i in 1:n_iterations) {
  t0 <- Sys.time()
  A <- matrix(rnorm(matrix_size * matrix_size), nrow = matrix_size)
  res <- svd(A)
  elapsed <- as.numeric(difftime(Sys.time(), t0, units = 'secs')) * 1000
  logs[[i]] <- data.frame(iteration = i, elapsed_ms = round(elapsed, 2))
}

benchmarks <- do.call(rbind, logs)
print(benchmarks)
```

```r
# 2. Save Benchmark Artifact & Publish provBook
saveRDS(benchmarks, 'hpc_benchmarks.rds')
# provBookR::publish_codex('notebooks/02_hpc_development_story.R', output_dir = 'books/hpc_development_story')
```
