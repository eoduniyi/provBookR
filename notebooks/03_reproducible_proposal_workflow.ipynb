# Showcase Notebook 03: Collaborative Proposal & Artifact Publication
Demonstrates multi-reviewer synthesis, proposal drafting, MD5 artifact checksum verification, and `provBookR` publication.

```r
# 1. Draft Proposal Content and Reviews
title <- 'Interactive Computational Provenance in High-Performance Computing'
intro <- 'Computational experiments require fine-grained provenance to ensure reproducibility.'
reviewers <- c(r1 = 'Dr. Alice Vance (MIT)', r2 = 'Dr. Bob Miller (Harvard)')
reviews <- c(r1 = 'Approved with minor edits.', r2 = 'Methodology is scalable.')

proposal_doc <- paste(
  paste0('# ', title),
  '## Introduction', intro,
  '## Peer Review Sign-off',
  paste('-', reviewers, ':', reviews, collapse = '\n'),
  sep = '\n\n'
)

writeLines(proposal_doc, 'reproducible_proposal.md')
tools::md5sum('reproducible_proposal.md')
```
