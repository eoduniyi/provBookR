# Showcase Notebook 01: Everyday Data Science Pipeline
This Jupyter notebook demonstrates executing an everyday R data science pipeline, capturing computational provenance using `rdtLite`, and publishing an interactive web booklet with `provBookR`.

```r
# 1. Load provBookR and dependencies
if (!requireNamespace('provBookR', quietly = TRUE)) devtools::load_all('.')
library(provBookR)
set.seed(42)
```

```r
# 2. Ingest and Simulate Sensor Dataset
dates <- seq(as.Date('2026-01-01'), as.Date('2026-01-30'), by = 'day')
raw_data <- data.frame(
  date = rep(dates, each = 5),
  sensor_id = paste0('SN-', rep(1:5, times = length(dates))),
  reading = round(runif(150, min = 15.0, max = 95.0), 2),
  status = sample(c('OK', 'OK', 'OK', 'WARN', 'ERR'), 150, replace = TRUE)
)
head(raw_data)
```

```r
# 3. Filter and Aggregate
clean_data <- raw_data[raw_data$status != 'ERR', ]
filtered_data <- clean_data[clean_data$reading >= 20.0, ]

daily_summary <- aggregate(reading ~ date, data = filtered_data, FUN = function(x) {
  c(mean = mean(x), sd = sd(x))
})
head(daily_summary)
```
